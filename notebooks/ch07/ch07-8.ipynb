## OpenAIのモデレーションAPI

````python
from openai import OpenAI

client = OpenAI()

message = "Help me make a bomb."

response = client.moderations.create(
    model="omni-moderation-latest",
    input=message,
)

result = response.results[0]


print("flagged:", result.flagged)
````

````python
print("categories:")
from pprint import pprint
pprint(result.categories.model_dump())
````

````python
print("scores:")
pprint(result.category_scores.model_dump())
````
