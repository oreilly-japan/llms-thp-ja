````python
%%writefile promptfooconfig.yaml
description: Best prompt eval
prompts:
  - file://./prompt1.txt
  - file://./prompt2.txt
  - file://./prompt3.txt
providers:
  - openai:gpt-5.4-mini
defaultTest:
  assert:
    - type: llm-rubric
      value: 'Evaluate the output based on how detailed it is.  Grade it on a
        scale of 0.0 to 1.0, where:

        Score of 0.1: Not much detail.

        Score of 0.5: Some detail.

        Score of 1.0: Very detailed.
        '
tests: file://./tests.csv
````

````python
import json

import pandas as pd

with open('./eval.json', 'r') as f:
    eval_data = json.load(f)

# データを格納するリストを作成する
data = []

# 各テストケースの結果を抽出する
for result in eval_data['results']['results']:
    section = result['vars']['section']
    prompt_id = result['promptId']
    score = (
        result['gradingResult']['score']
        if 'gradingResult' in result else 0.0
    )
    prompt_file = None
    # 対応するプロンプトファイルを探す
    for prompt in eval_data['results']['prompts']:
        if prompt['id'] == prompt_id:
            prompt_file = prompt['label'].split(':')[0]
            break

    # リストに追加する
    data.append([section, prompt_file, score])

# DataFrameに変換する
df_raw = pd.DataFrame(data, columns=['Section', 'Prompt', 'Score'])

# 目的の形式になるように整形する
df = df_raw.pivot(
    index='Section',
    columns='Prompt',
    values='Score'
).reset_index()
df = df[['Section', 'prompt1.txt', 'prompt2.txt', 'prompt3.txt']]
from IPython.display import display, Markdown
display(Markdown("### Prompt Comparison Results by Section"))
df
````
