````python
%%writefile prompt1.txt
- role: system
  content: "You are an expert financial analyst. Write concise, accurate summaries."
- role: user
  content: "Generate a 1-line summary of the section '{{section}}' from an SEC filing.\n\nSection content:\n{{content}}"
````

````python
%%writefile promptfooconfig.yaml
description: Best model eval
prompts:
- file://./prompt1.txt
providers:
- openai:gpt-5.4
- openai:gpt-5.4-mini
- openai:gpt-4.1-mini
defaultTest:
  assert:
  - type: cost
    threshold: 0.003
  - type: latency
    threshold: 3000
  - type: python
    value: len(output) < 3000
  - type: llm-rubric
    value: Does the summary look like it was written by an expert analyst
      [Yes/No]?
tests: file://./tests.csv
````

test.csvアップロード

````python
!promptfoo eval --no-cache --output eval.json
````

````python
!head -n 20 eval.json
````

````python
import json
import pandas as pd

# JSONファイルを読み込む
with open('./eval.json', 'r') as f:
    eval_data = json.load(f)

# 結果を辞書のリストに格納する
results = []
for prompt in eval_data['results']['prompts']:
    result = {
        'model': prompt['provider'],
        'latency_ms': prompt['metrics']['totalLatencyMs'],
        'token_usage': prompt['metrics']['tokenUsage']['total'],
        'cost': prompt['metrics']['cost'],
        'assert_pass': prompt['metrics']['assertPassCount'],
        'assert_fail': prompt['metrics']['assertFailCount'],
        'prompt_tokens': prompt['metrics']['tokenUsage']['prompt'],
        'num_requests': prompt['metrics']['tokenUsage']['numRequests']
    }
    results.append(result)

from IPython.display import display, Markdown

# DataFrameに変換する
df = pd.DataFrame(results)
````
